## **Manually generating histogram**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#  calculate number of bins required:

def get_bins(x):
  return int(np.ceil(np.log(len(x))/np.log(2)))

def get_range(x):
  return (x.max()-x.min())

def get_classWidth(x):
  ranges = get_range(x)
  bins = get_bins(x)
  width = np.ceil(ranges/bins)
  return int(width)

def get_classRange(x):
  width = get_classWidth(x)
  bins = get_bins(x)
  # print(width,bins)
  x.sort()
  intervals = np.arange(x[0],x[-1],width)
  l = []
  j = 0
  for i in range(bins):
    k = [intervals[j],intervals[j]+width]
    j = j+1
    l.append(k)
  return l

def get_frequency(x):
  counts = []
  intervals = get_classRange(x)
  for r in intervals:
    freq = ((r[0] <= x) & (x < r[1])).sum()
    counts.append(freq)
  return counts



```
