# What is Tensor ?

- Tensors are multi-dimensional arrays with a uniform type (called a dtype). 

- If you're familiar with NumPy, tensors are (kind of) like np.arrays.

- All tensors are immutable like Python numbers and strings: you can never update the contents of a tensor, only create a new one.

```python
import tensorflow as tf
import numpy as np
```


---
**"scalar" or "rank-0" tensor . A scalar contains a single value, and no "axes"**


```python
rank_0_tensor = tf.constant(4)
```

```python
print(rank_0_tensor)
```

**A "vector" or "rank-1" tensor is like a list of values. A vector has one axis:**

```python
rank_1_tensor = tf.constant([2.0,3.0,4.0,5.0])
```

```python
print(rank_1_tensor)
```

**A "matrix" or "rank-2" tensor has two axes:**

```python
rank_2_tensor = tf.constant([
                             [1,2],
                             [3,4],
                             [5,6]
],dtype=np.float32)
```

```python
print(rank_2_tensor)
```

**Tensors may have more axes; here is a tensor with three axes:**

```python
x = np.arange(30).reshape(3,2,5)
```

```python
rank_3_tensor = tf.constant(x,dtype=np.float32)
```

```python
print(rank_3_tensor)
```

You can do basic math on tensors, including addition, element-wise multiplication, and matrix multiplication.

```python
a = tf.constant([[1,2],[3,4]])
b = tf.constant([[1,1],[1,1]])
```

```python
tf.add(a,b)
```

```python
tf.subtract(a,b)
```

```python
tf.multiply(a,b)
```

```python
tf.matmul(a,b)
```

or simply,

```python
a+b
```

```python
a-b
```

```python
a*b #element wise multiplication
```

```python
a @ b  # matrix multiplication
```

# Tensorflow Variable

- A TensorFlow variable is the recommended way to represent shared, persistent state your program manipulates. 

- Variables are created and tracked via the tf.Variable class. 

- A tf.Variable represents a tensor whose value can be changed by running ops on it. 

- Specific ops allow you to read and modify the values of this tensor. Higher level libraries like tf.keras use tf.Variable to store model parameters.

---
# Create a variable

To create a variable, provide an initial value. The tf.Variable will have the same dtype as the initialization value.


```python
x = tf.constant([1.0,2.0,3.0,4.0])
y = tf.Variable(x)
```

```python
x.dtype
```

```python
y.dtype
```

```python
print(type(x))
print(type(y))
```

```python
a = tf.Variable([1.0,2.0])
b = tf.Variable(a)
```

```python
a.assign([5,6])
```

```python
a.numpy()
```

```python
b.numpy()
```
