<h1>Apply Functions</h1>

```python
df = pd.DataFrame(data=np.arange(12).reshape(3,4),columns=list("ABCD"))
```

```python
df
```

```python
df.apply(np.sqrt)
```

```python
df.apply(np.cumsum)
```

```python
df.apply(np.cumsum,axis=1)
```

```python
df.apply(lambda i: i+2)
```

```python
df.apply(lambda i: i+2,axis=0)
```

```python
df.apply(lambda i:i%2==0)
```

```python
def test(x):
    return list(filter(lambda i:i%2==0,x))
```

```python
df.apply(test)
```

```python
def get_mean(x):
    return np.mean(x)
```

```python
df.apply(get_mean)
```

```python
df.apply(np.mean)
```

<h1>Join</h1>

```python
df1 = pd.DataFrame({
    "fruits":["Apple","Mango","Orange"],
    "Price":[100,200,89]
})
```

```python
df2 = pd.DataFrame({
    "fruits":["Apple","Mango","Orange","Banana"],
    "Price":[100,200,30,45]
})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="fruits")
```

```python
pd.merge(df1,df2)
```

```python
pd.merge(df1,df2,how="outer")
```

```python
pd.merge(df1,df2,how="left")
```

```python
pd.merge(df1,df2,how="right")
```

Grouping

By “group by” we are referring to a process involving one or more of the following steps:

- Splitting the data into groups based on some criteria

- Applying a function to each group independently

- Combining the results into a data structure

```python
df = pd.DataFrame({
    'A':["foo","bar"]*4,
    'B':["one", "one", "two", "three", "two", "two", "one", "three"],
    'C':np.random.randn(8),
    'D':np.random.randn(8)
})
```

```python
df
```

```python
list(df.groupby("A"))
```

```python
df.groupby("A").sum()
```

```python
df.groupby("B").sum()
```

```python
df.groupby("B").sum().T
```
