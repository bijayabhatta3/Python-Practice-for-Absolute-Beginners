Missing data


pandas primarily uses the value np.nan to represent missing data. It is by default not included in computations.

Reindexing allows you to change/add/delete the index on a specified axis. This returns a copy of the data:

```python
df
```

```python
df3 = df.reindex(index=pd.date_range("2022-01-20",periods=4),columns=list(df.columns)+["E"])
```

```python
df3
```

```python
df3.loc[:2,"E"] = 1
```

```python
df3
```

To drop any rows that have missing data:

```python
df3.isna()
```

```python
df3.isna().sum()
```

```python
df3.dropna(how='any')
```

```python
df3
```

```python
df3.dropna(axis=0)
```

```python
df3.dropna(axis=1)
```

```python
df3.fillna(value=np.mean(df3['E']))
```

Join

SQL style merges.

```python
d1 = pd.DataFrame({
    "Product":["Apple","Oranges"],
    "Prices":[10,11]
})
```

```python
d1
```

```python
d2 = pd.DataFrame({
    "Product":["Apple","Oranges","Mango"],
    "Prices":[10,11,12]
})
```

```python
d2
```

```python
pd.merge(d1,d2,how="inner",on="Product")
```

```python
pd.merge(d1,d2,how="outer",on="Product")
```
