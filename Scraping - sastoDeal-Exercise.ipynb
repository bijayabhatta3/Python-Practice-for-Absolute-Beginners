```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.sastodeal.com/sd-fast/food-essentials/rice-rice-products.html?p=1"
```

```python
raw_html = requests.get(url)
```

```python
# print(raw_html.text)
```

```python
soup = BeautifulSoup(raw_html.text,'html.parser')
```

```python
# class="products list items product-items"
raw_soup = soup.find("ol",class_=["products","list","items","product-items"])
```

```python
raw_soup
```

```python
raw_data = raw_soup.find_all('li')
```

```python
first_row = raw_data[0]
```

```python
first_row
```

```python
first_row.strong.text
```

```python
prices = first_row.find_all('span',attrs={'class':'price'})
```

```python
prices[0].text
```

```python
prices[1].text
```

```python
master = []
for data in raw_data:
  d = {}
  d['name']  = data.strong.text
  prices = data.find_all('span',attrs={'class':'price'})
  if(len(prices)==2):
    d['old_price'] = prices[0].text
    d['new_price'] = prices[1].text
  else:
    d['price'] = prices[0].text
  master.append(d)
```
