```python
url = "https://www.sastodeal.com/sd-fast/food-essentials/rice-rice-products.html?p={}"
```

```python
def get_html_doc(url):
  response = requests.get(url)
  return (response.text)
```

```python
def parse_html(html_text):
  soup = BeautifulSoup(html_text,"html.parser")
  return soup
```

```python
def get_data(soup):
  data = []
  contents = soup.find("ol",class_=["products","list","items","product-items"])
  for content in contents.children:
    if content!=' ':
      d = {}
      product_details = content.find("div",class_=["product","details","product-item-details"])
      # print(product_details.find("strong").text)
      d['product_name'] = product_details.find("strong").text
      # print(product_details.find('a')['href'])
      d['product_url'] = product_details.find('a')['href']
      prices = product_details.find_all("span",attrs={'class':'price'})
      if(len(prices)==2):
        d['old_price'] = prices[0].text
        d['new_price'] = prices[1].text
      else:
        d['price'] = prices[0].text
      data.append(d)
  return data
```

```python
master_data = []
for i in range(1,5):
  
  url = 'https://www.sastodeal.com/sd-fast/food-essentials/rice-rice-products.html?p={}'.format(i)
  html_text = get_html_doc(url)
  soup = parse_html(html_text)
  master_data.extend(get_data(soup))

```
