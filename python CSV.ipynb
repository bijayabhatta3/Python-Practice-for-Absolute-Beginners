```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.sastodeal.com/sd-fast/food-essentials/rice-rice-products.html?p={}"
```

```python
def get_html_doc(url):
  response = requests.get(url)
  return (response.text)
```

```python
def parse_html(html_text):
  soup = BeautifulSoup(html_text,"html.parser")
  return soup
```

```python
def get_data(soup):
  data_list = []
  raw_soup = soup.find("ol",class_=["products","list","items","product-items"])
  raw_data = raw_soup.find_all('li')
  for data in raw_data:
    d = {}
    d['name']  = data.strong.text
    prices = data.find_all('span',attrs={'class':'price'})
    if(len(prices)==2):
      d['old_price'] = prices[0].text
      d['new_price'] = prices[1].text
    else:
      d['new_price'] = prices[0].text
      d['old_price'] = None

    data_list.append(d)
  return data_list


```

```python
master_data = []
for i in range(1,5):
  
  url = 'https://www.sastodeal.com/sd-fast/food-essentials/rice-rice-products.html?p={}'.format(i)
  html_text = get_html_doc(url)
  soup = parse_html(html_text)
  get_data(soup)
  master_data.extend(get_data(soup))

```

```python
fieldnames = list(master_data[0].keys())
print(fieldnames)
```

```python
with open("snapdeal.csv","w") as file:
  writer = csv.DictWriter(file,fieldnames=fieldnames)
  writer.writeheader()
  for data in master_data:
    writer.writerow(data)
```
