## **use the following code for reference:**

```python
import requests
from bs4 import BeautifulSoup
```

```python
html = requests.get('https://ekantipur.com/news')
```

```python
soup = BeautifulSoup(html.text)
```

```python
articles = soup.find_all('article',attrs={'class':'normal'})
articles
```

```python
article = articles[0]
article
```

```python
base_url = 'https://ekantipur.com/'
article_url = base_url+article.a['href']
```

```python
article_url
```

```python
article_title = article.a.text
```

```python
article_title
```

```python
article_author = article.find('div',attrs={'class':'author'}).text 
```

```python
article_author
```

```python
article_full = article.p.text
```

```python
article_full
```

```python
data=[]
for article in articles:
  d = {}
  d['article_url'] = base_url+article.a['href']
  d['title'] = article.a.text
  d['author'] = article.find('div',attrs={'class':'author'}).text
  d['news'] = article.p.text
  data.append(d)


```

```python
data
```

```python
import csv
```

```python
# f = open('test.txt','r')
with open("samachar.csv","w") as f:
  field_names = list(data[0].keys())
  writer = csv.DictWriter(f,fieldnames=field_names)
  writer.writeheader()
  for news in data:
    writer.writerow(news)
```
