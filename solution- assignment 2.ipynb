```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.pythonscraping.com/pages/page3.html"
```

```python
html_doc = requests.get(url)
```

```python
soup = BeautifulSoup(html_doc.text,"html.parser")
```

```python
table_raw = soup.find('table',attrs={'id':'giftList'})
```

```python
master = []
for row in table_raw.children:
  if(row!='\n'):
    tds = row.find_all('td')
    # print(tds)
    if tds:
      d = {}
      d['item_title'] = tds[0].text
      d['description'] = tds[1].text
      d['cost'] = tds[2].text
      d['link'] = tds[3].img['src']
    master.append(d)




```
