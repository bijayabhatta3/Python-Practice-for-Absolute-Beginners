```python
from bs4 import BeautifulSoup
```

```python
html_str = """
<!DOCTYPE html>
<html>
<head>
<title>learn css</title>

<style>
.test{
color:green;
}

#demo{
color:blue;
}

</style>

</head>

<body>
<h2 style="color:red;">Introduction</h2>

<p class="test">Hello welcome to python class</p>

<p class="test">we will learn python and its libraries</p>

<p> we will learn following: </p>

<ul id="demo">
<li>JAVA</li>
<li>Python</li>
<ul>

</body>

</html>
"""
```

```python
print(html_str)
```

```python
soup = BeautifulSoup(html_str,"html.parser")
```

```python
print(soup)
```

```python
lines = soup.find_all("p",attrs={"class":"test"})
```

```python
for line in lines:
  print(line.text)
```

```python
lists = soup.find("ul",attrs={"id":"demo"})
```

```python
lists
```

```python
li_info = lists.find_all("li")
```

```python
li_info
for i in li_info:
  print(i.text)
```

```python
lists
```

```python
list(lists.children)
```

```python
for child in lists.children:
  print(child)
```

```python
d = lists.li.next_sibling.next_sibling
```

```python
d.previous_sibling.previous_sibling
```
